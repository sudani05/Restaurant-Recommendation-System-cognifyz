# Restaurant Recommendation System

## Objective
The objective of this project is to build a restaurant recommendation system that suggests restaurants based on user preferences such as cuisine, city, and price range.

### Techniques Used
- Content-Based Filtering
- CountVectorizer
- Cosine Similarity

### Expected Output
The system recommends restaurants that are similar to the user's preferences.

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("Dataset .csv")
```

```python
df["Cuisines"] = df["Cuisines"].fillna("Unknown")
```

```python
df["City"] = df["City"].fillna("Unknown")
```

```python
data = df[[
    "Restaurant Name",
    "Cuisines",
    "Average Cost for two",
    "Price range",
    "Aggregate rating",
    "City"
]].copy()
```

```python
data.head()
```

```python
data["features"] = (
    data["Cuisines"].astype(str) + " " +
    data["City"].astype(str) + " " +
    data["Price range"].astype(str)
)
```

```python
data[["Restaurant Name", "features"]].head()
```

```python
cv = CountVectorizer(stop_words="english")
```

```python
matrix = cv.fit_transform(data["features"])
```

```python
similarity = cosine_similarity(matrix)
```

```python
similarity.shape
```

```python
def recommend_restaurant(cuisine, city, price_range):

    
    filtered_data = data[
        (data["City"].str.lower() == city.lower()) &
        (data["Price range"] == price_range)
    ]

    
    if filtered_data.empty:
        return "No restaurants found matching your criteria."


    filtered_features = filtered_data["features"]


    filtered_matrix = cv.transform(filtered_features)

    user_input = cuisine + " " + city + " " + str(price_range)


    user_vector = cv.transform([user_input])

    scores = cosine_similarity(user_vector, filtered_matrix)

    scores = scores.flatten()


    top_indexes = scores.argsort()[::-1][:10]

    
    recommendations = filtered_data.iloc[top_indexes][[
        "Restaurant Name",
        "Cuisines",
        "City",
        "Price range",
        "Aggregate rating"
    ]]

    recommendations = recommendations[
        recommendations["Aggregate rating"] > 3.5
    ]

    recommendations = recommendations.drop_duplicates(
        subset="Restaurant Name"
    )

    
    recommendations = recommendations.sort_values(
        by="Aggregate rating",
        ascending=False
    )

    return recommendations
```

```python
recommend_restaurant(
    cuisine="North Indian",
    city="New Delhi",
    price_range=2
)
```

```python
recommend_restaurant(
    cuisine="Chinese",
    city="Mumbai",
    price_range=3
)
```

```python
recommend_restaurant(
    cuisine="Fast Food",
    city="Noida",
    price_range=2
)
```

```python
cuisine = input("Enter Cuisine: ")
city = input("Enter City: ")
price = int(input("Enter Price Range (1-4): "))

recommend_restaurant(cuisine, city, price)
```

# Conclusion

A content-based restaurant recommendation system was successfully developed using restaurant features such as cuisine, city, and price range.

CountVectorizer was used for text vectorization, while Cosine Similarity was used to identify similar restaurants. The system provides personalized restaurant recommendations based on user preferences.
